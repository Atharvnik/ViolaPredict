```python

import pandas as pd
import numpy as np

df = pd.read_csv(
    r"C:\Users\HP\Downloads\mini projectnew2\Data\Traffic_Violations.csv",
    low_memory=False
)

print(df.shape)
```

```python
df.head(10)
```

```python
df = df.sample(n=300000, random_state=42)

print(df.shape)
```

```python
df.columns = (
    df.columns
      .str.strip()
    .str.lower()# remove extra spaces
      .str.replace(" ", "_")    # spaces → underscore
      .str.replace("/", "_")    # slash → underscore
      .str.replace("-", "_")    # dash → underscore
)
```

```python
df.columns
```

```python
df = df.rename(columns={
    "vehicletype": "vehicle_type",
    "date_of_stop": "stop_date",
    "time_of_stop": "stop_time"
})
```

```python
df.columns
```

```python
df.isna().sum().sort_values(ascending=False).head(10)
```

```python
df["latitude"] = df["latitude"].fillna(df["latitude"].median())
df["longitude"] = df["longitude"].fillna(df["longitude"].median())
```

```python
df["geolocation"] = df["geolocation"].fillna("Unknown")
```

```python
categorical_nan_cols = [
    "article",
    "color",
    "dl_state",
    "driver_city",
    "model",
    "make"
]

df[categorical_nan_cols] = df[categorical_nan_cols].fillna("Unknown")
```

```python
df["year"] = df["year"].fillna(df["year"].median()).astype(int)
```

```python
df.isna().sum().sort_values(ascending=False).head(10)
```

```python
remaining_object_nan_cols = [
    "state",
    "driver_state",
    "description",
    "subagency"
]
```

```python
df[remaining_object_nan_cols] = df[remaining_object_nan_cols].fillna("Unknown")
```

```python
df.isna().sum().sort_values(ascending=False).head(10)
```

```python
# Select object columns
object_cols = df.select_dtypes(include="object").columns

# Count empty or whitespace-only values
empty_counts = {}

for col in object_cols:
    empty_counts[col] = ((df[col].str.strip() == "")).sum()

# Show columns with empty values
pd.Series(empty_counts).sort_values(ascending=False).head(10)
```

```python
df.head(5)
```

```python
df["stop_date"] = pd.to_datetime(df["stop_date"], errors="coerce")
```

```python
df["stop_time"] = pd.to_datetime(df["stop_time"], format="%H:%M:%S", errors="coerce").dt.time
```

```python
df[["stop_date", "stop_time"]].dtypes
```

```python
df["stop_hour"] = df["stop_time"].apply(
    lambda x: x.hour if pd.notnull(x) else -1
)

df["stop_day"] = df["stop_date"].dt.day
df["stop_month"] = df["stop_date"].dt.month
df["stop_weekday"] = df["stop_date"].dt.weekday
```

```python
# Total duplicate rows
df.duplicated().sum()
```

```python
df = df.drop_duplicates()
```

```python
df.to_csv(
    r"C:\Users\HP\Downloads\mini projectnew2\Data\cleaned_stop_level_dataset.csv",
    index=False
)

print("✅ Stop-level cleaned dataset saved for visualization")
```

```python
df["vehicle_id"] = (
    df["make"].astype(str) + "_" +
    df["model"].astype(str) + "_" +
    df["year"].astype(str)
)
```

```python
vehicle_df = df.groupby("vehicle_id").agg(
    total_violations=("vehicle_id", "count"),
    accident_count=("accident", lambda x: (x == "Yes").sum()),
    alcohol_count=("alcohol", lambda x: (x == "Yes").sum()),
    fatal_count=("fatal", lambda x: (x == "Yes").sum()),
    property_damage_count=("property_damage", lambda x: (x == "Yes").sum()),
    avg_stop_hour=("stop_hour", "mean"),
    unique_locations=("location", "nunique")
).reset_index()
```

```python
threshold = vehicle_df["total_violations"].quantile(0.90)

vehicle_df["repeat_risk"] = (
    vehicle_df["total_violations"] >= threshold
).astype(int)
```
